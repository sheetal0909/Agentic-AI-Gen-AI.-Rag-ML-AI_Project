```python
num1 = 15
num2 = 28
num3 = 21
num4 = 30
result = num1 + num2
add = num1 + num2 + num3 + num4
print('The Addiltion of {} and {} is = {}'.format(num1,num2,result))
print('The Addiltion of {} and {} and {} and {} is = {}'.format(num1,num2,num3,num4,add))
print(f'The Addiltion of {num1} and {num2} is = {result}') # F String (Short Form of F String)
```
