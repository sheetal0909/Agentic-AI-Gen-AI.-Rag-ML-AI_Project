8.Write a Python program that accepts a string and prints the 
number of vowels in it. 

```python
text = input("Enter a string: ")

vowels = "aeiou"
count = 0

for char in text.lower():
    if char in vowels:
        count += 1

print("Number of vowels:", count)
```

16.Write a program that checks if the input number is a prime number

```python
num = int(input("Enter a number: "))
if num > 1:
    print("Not a prime number")
else:
    for i in range(2, num):
        if num % i == 0:
            print("Not a prime number")
            break
    else:
        print("Prime number")
```

 48. Write a program to count the number of consonants in a string entered by the user. 

```python
string_input = input("Enter a string: ") 
consonants = "bcdfghjklmnpqrstvwxyz" 
count = sum(1 for char in string_input.lower() if char in consonants) 
print("Number of consonants:", count)
```
