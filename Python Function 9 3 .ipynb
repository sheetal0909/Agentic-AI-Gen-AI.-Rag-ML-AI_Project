# Type of Funcyional Arguments

 1 - Positional Argument

```python
def person(name,age):
    print(name)
    print(age)
person("NIT" ,24)
```

```python
def person(name,age):
    print(name)
    print(age)
person("NIT")
```

```python
def person(name,age):
    print(name)
    print(age)
person(23,"NIT")
```

```python
def person(name,age):
    print(name)
    print(age+1)
person("NIT",24)
```

```python
def person(name,age):
    print(name)
    print(age+1)
person(24,"NIT")
```

2 - Keyboard Arguments

```python
def person(name,age):
    print(name)
    print(age+1)
person(name = "NIT" , age = 24) #Declare like this to avoid positional error
```

```python
def person(name,age,mobile):
    print(name)
    print(age+1)
person(name = "NIT" , age = 24,1234)
```

```python
def person(name,age,mobile):
    print(name)
    print(age+1)
person(name = "NIT" , age = 24,mobile=1234)
```

```python
def person(name,age,mobile):
    print(name)
    print(age+1)
person(name = "NIT" , age = 24,mobile=1234, address = "HyD")
```

```python
def person(name,age,mobile,address):
    print(name)
    print(age+1)
person(name = "NIT" , age = 24,mobile=1234, address = "HyD")
```

3 - Default Argument

```python
def person(name,age = 18):
    print(name)
    print(age)
person('vote')
```
