```python
# Pyramid Pattern
for i in range(1,6):
    print(' '*(5-i)+' * '*(2*i-1))
```

```python
 # Inverted Pyramid
for i in range(5,0,-1):
    print(' '*(5-i)+' * '*(2*i-1))
```

```python
# Dimaond Pyramid
for i in range(1,6):
    print(' '*(5-i)+' * '*(2*i-1))
for i in range(4,0,-1):
    print(' '*(5-i)+' * '*(2*i-1))
```

```python
# Hallow Square pattern
for i in range(5):
    for j in range(5):
        if i == 0 or i == 4 or j == 0 or j == 4:
            print('*' ,end='')
        else:   
            print(' ' ,end='')
    print()         
```
