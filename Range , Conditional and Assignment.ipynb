Question 8:  Write a Python program that accepts a string and prints the 
number of vowels in it.

```python
name = input("Enter a Name ")
vowels = "aeiou"
count = sum(1 for char in name if char.lower() in vowels)
print("Number of vowels:", count)
```

Question 10: How would you check if a string is a palindrome using `input()`? 

```python
text = input("Enter a string: ") 
if text == text[::-1]: 
    print("Palindrome") 
else: 
    print("Not a palindrome") 
```
